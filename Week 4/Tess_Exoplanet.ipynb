```python
import lightkurve as lk
import matplotlib.pyplot as plt
import numpy as np
from astropy import units as u

%matplotlib inline
```

```python
#Yıldız verisini aratalım -> TOI-700 yıldızının Tess gözlem verileri
search_result = lk.search_lightcurve("TOI-700", mission="TESS", author="SPOC")
print(search_result)
```

```python
# İlk gözlem verisini indiriyoruz (download komutu burada çalışıyor)
lc = search_result[0].download()

# Işık eğrisini çizdiriyoruz
lc.select_flux("pdcsap_flux").plot()
plt.savefig("light_curve.png", dpi=300, bbox_inches="tight")
plt.show()
```
